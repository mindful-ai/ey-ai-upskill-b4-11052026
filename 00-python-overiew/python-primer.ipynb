### Branching

```python
a = 10
b = 1
res = a - b
if res > 0:
    print("result is positive")
elif res < 0:
    print("result is negative")
else:
    print("result is zero")
```

```python
match res:
    case res if res > 0:    print("result is positive")
    case res if res < 0:    print("result is negative") 
    case _:                print("result is zero")
```

### Example: Program to determine if a number is prime or not

```python
N = 12

prime = True
for i in range(2, N):
    if N % i == 0:
        prime = False
        break

if prime:
    print(N, "is a prime number")
else:
    print(N, "is not a prime number")
```

```python
def is_prime(N):
    if N < 2:
        return False
    for i in range(2, N):
        if N % i == 0:
            return False
    return True
```

```python
print("List of prime numbers between 1 and 100:")
for num in range(2, 101):
    if is_prime(num):
        print(num, end=" ")
```
